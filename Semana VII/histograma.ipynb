## Ecualización del histograma

```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv.imread('/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/squirrel_cls.jpg', 0)
```

```python
img.shape
```

```python
plt.imshow(img, 'gray')
plt.show()
```

```python
# Esto NO se hace
histograma_org = cv.calcHist([img], (0,), None, [256], (0, 255))
plt.plot(histograma_org)
plt.show()
```

```python
# Ecualizar el hist
img_eq = cv.equalizeHist(img)
plt.plot(cv.calcHist([img_eq], (0,), None, [256], (0, 255)))
plt.show()
```

```python
plt.imshow(np.hstack((img, img_eq)), 'gray')
plt.show()
```

```python
img_color = cv.imread('/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/messi5.jpg')[:, :, ::-1]
```

```python
color = ["r", "g", "b"]

for contador, valor in enumerate(color):
    # print(i)
    tmp = cv.calcHist([img_color], (contador,), None, [256], (0, 255))
    plt.plot(tmp, valor)
    plt.title(f"Histograma de los canales")
plt.show()
```

```python
r, g, b = cv.split(img_color)
r_eq = cv.equalizeHist(r)
img_r_eq = cv.merge((r_eq, g, b))
for contador, valor in enumerate(color):
    # print(i)
    tmp = cv.calcHist([img_r_eq], (contador,), None, [256], (0, 255))
    plt.plot(tmp, valor)
    plt.title(f"Histograma de los canales")
plt.show()
```

```python
plt.imshow(np.hstack((img_color, img_r_eq)))
plt.show()
```

```python
g_eq = cv.equalizeHist(g)
b_eq = cv.equalizeHist(b)
img_color_eq = cv.merge((r_eq, g_eq, b_eq))
for contador, valor in enumerate(color):
    # print(i)
    tmp = cv.calcHist([img_color_eq], (contador,), None, [256], (0, 255))
    plt.plot(tmp, valor)
    plt.title(f"Histograma de los canales")
plt.show()
```

```python
plt.imshow(img_color_eq)
plt.show()
```
