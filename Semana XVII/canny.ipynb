```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/cards.png", 0)
```

```python
img_blur = cv.blur(img, (3, 3))
```

```python
plt.imshow(img_blur, "gray")
plt.show()
```

```python
# cv.Sobel()

low_threshold = 100
ratio = 2

img_bordes = cv.Canny(img_blur, low_threshold, ratio*low_threshold, 3)
plt.figure()
plt.subplot(121)
plt.imshow(img_blur, "gray")
plt.title("Img. orig.")
plt.axis("off")
plt.subplot(122)
plt.imshow(img_bordes, "gray")
plt.title("Img. Canny")
plt.axis("off")
plt.show()
```
