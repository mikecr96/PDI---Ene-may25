```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread('/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/aloeL.jpg', 0)
plt.imshow(img, "gray")
plt.show()
```

```python
img_bin = cv.threshold(img, 127, 255, cv.THRESH_BINARY)[1]
plt.imshow(img_bin, 'gray')
plt.show()
```

```python
s_e = np.ones((3,3) , np.uint8)
```

```python
s_e
```

```python
img_bin_erode = cv.erode(img_bin, s_e, iterations=2)
plt.imshow(img_bin_erode, 'gray')
plt.show()
```
