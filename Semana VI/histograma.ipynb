```python
img_color = cv.imread('/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/lena.jpg')[:, :, ::-1]
```

```python
plt.imshow(img_color)
plt.show()
```

```python
histograma_r = cv.calcHist([img_color], (0,), None, [256], (0, 255))
plt.plot(histograma_r, "r")
plt.title("Histograma del canal R")
plt.show()
```

```python
histograma_g = cv.calcHist([img_color], (1,), None, [256], (0, 255))
plt.plot(histograma_g, "g")
plt.title("Histograma del canal G")
plt.show()
```

```python
histograma_b = cv.calcHist([img_color], (2,), None, [256], (0, 255))
plt.plot(histograma_b, "b")
plt.title("Histograma del canal B")
plt.show()
```

```python
color = ["r", "g", "b"]

for contador, valor in enumerate(color):
    # print(i)
    tmp = cv.calcHist([img_color], (contador,), None, [256], (0, 255))
    plt.plot(tmp, valor)
    plt.title(f"Histograma de los canales")
plt.show()
```
