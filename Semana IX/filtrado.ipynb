```python
# ESTO SÍ SE HACE

def getKernel(n: int):
    return 1/(n**2) * np.ones((n, n), dtype=np.float32)
```

```python
getKernel(3).sum()
```

```python
getKernel(5).sum()
```

```python
getKernel(7).sum()
```

```python
# (inial; final; i++)
list(range(3, 28, 2))
```

```python
for i in range(3, 28, 2):
    img_filtrada = cv.filter2D(chicky_gray, -1, getKernel(i))
    plt.imshow(img_filtrada, "gray")
    plt.title(f"Tamaño del karnel: {i}")
    plt.show()
```
