```python
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt
```

```python
img = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/HappyFish.jpg", 0)
```

```python
img.shape
```

```python
img.dtype
```

```python
kernel1 = np.array([[0, 1, 0],
                    [1, -4, 1],
                    [0, 1, 0]])
```

```python
kernel2 = kernel1 * -1
```

```python
kernel3 = np.array([[1, 1, 1],
                    [1, -8, 1],
                    [1, 1, 1]])
kernel4 = kernel3 * -1
```

```python
print(kernel1.sum())
print(kernel2.sum())
print(kernel3.sum())
print(kernel4.sum())
```

```python
img_f1 = cv.filter2D(img, -1, kernel1)
```

```python
plt.imshow(img_f1, "gray")
plt.show()
```

```python
np.unique(img_f1)
```

```python
img_f2 = cv.filter2D(img, -1, kernel2)
plt.imshow(img_f2, "gray")
plt.axis("off")
plt.show()
```

```python
img_f3 = cv.filter2D(img, -1, kernel3)
plt.imshow(img_f3, "gray")
plt.axis("off")
plt.show()
```

```python
img_f4 = cv.filter2D(img, -1, kernel4)
plt.imshow(img_f4, "gray")
plt.axis("off")
plt.show()
```
