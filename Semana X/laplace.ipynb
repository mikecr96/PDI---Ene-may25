```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
src = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/jersey_diablos.png", 0)
```

```python
plt.imshow(src, "gray")
plt.show()
```

```python
# Preprocesamiento
src_denoised = cv.GaussianBlur(src, (3, 3), 0)
```

```python
# Calculamos Laplaciano
src_laplacian = cv.Laplacian(src_denoised, -1, ksize=3)
```

```python
plt.imshow(src_laplacian, "gray")
plt.show()
```
