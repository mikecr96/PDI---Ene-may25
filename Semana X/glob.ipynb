```python
# %pip install tqdm -U
```

```python
from tqdm import tqdm
```

```python
%matplotlib inline
```

```python
from time import time
```

```python
time()
```

```python
inicio = time()

imgs = []

for ruta in tqdm(all_images, unit="rutas", desc="Abriendo"):
    # print(f"{c+1}/{len(all_images)}")
    tmp = cv.imread(ruta, 0)
    imgs.append(tmp)
    plt.imshow(tmp, "gray")
    plt.title(ruta)
    plt.axis("off")
    plt.show()

fin = time()

print(f"Tiempo transcurrido: {fin - inicio:.3f} segundos.")
```
