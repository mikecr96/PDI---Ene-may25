```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/baboon.jpg",  0)
```

```python
histograma = cv.calcHist([img], [0], None, [256], [0, 256])
```

```python
type(histograma)
```

```python
histograma.shape
```

```python
plt.plot(histograma)
plt.show()
```

```python
img_color = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/baboon.jpg")[:, :, ::-1]
canalR = cv.calcHist([img_color], [0], None, [256], [0, 256])
canalG = cv.calcHist([img_color], [1], None, [256], [0, 256])
canalB = cv.calcHist([img_color], [2], None, [256], [0, 256])
```

```python
colores = ["r", "g", "b"]

for i, color in enumerate(colores):
    tmp = cv.calcHist([img_color], [i], None, [256], [0, 256])
    plt.plot(tmp, color)
    plt.title(f"Histograma del color {color}")
plt.show()
```

Ecualización del hist

```python
babuino_eq = cv.equalizeHist(img)
```

```python
plt.imshow(np.hstack((img, babuino_eq)), "gray")
plt.show()
```

```python
histograma_eq = cv.calcHist([babuino_eq], [0], None, [256], [0, 256])
plt.plot(histograma)
plt.plot(histograma_eq)
plt.show()
```
