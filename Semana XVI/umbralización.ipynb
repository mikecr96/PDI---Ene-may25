```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

---
Umbralización GLOBAL
1) Función de OpenCV

```python
sudoku_gray = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/sudoku.png", 0)
plt.imshow(sudoku_gray, "gray")
plt.show()
```

```python
sudoku_bin = cv.threshold(sudoku_gray, 50, 255, cv.THRESH_BINARY)[1]
```

```python
plt.imshow(sudoku_bin, "gray")
plt.show()
```

---
Función de numpy

```python
sudoku_bin = np.where(sudoku_gray >= 50, 255, 0)
plt.imshow(sudoku_bin, "gray")
plt.show()
```

---
2) Umbralización GLOBAL + mejora

```python
umbral, sudoku_bin = cv.threshold(sudoku_gray, 0, 255, cv.THRESH_BINARY | cv.THRESH_OTSU)
plt.imshow(sudoku_bin, "gray")
plt.title(f"Umbra OTSU: {umbral}")
plt.show()
```

```python
umbral, sudoku_bin = cv.threshold(sudoku_gray, 1000000000, 255, cv.THRESH_BINARY + cv.THRESH_TRIANGLE)
plt.imshow(sudoku_bin, "gray")
plt.title(f"Umbra TRIANGLE: {umbral}")
plt.show()
```

---
3) Umbralización ADAPTATIVA

```python
sudoku_gray = cv.imread("/Users/miguelcamargorojas/Documents/UP/PDI - Ene-may25/images/sudoku.png", 0)
sudoku_gray = cv.medianBlur(sudoku_gray, 5) # PRE-PROCESAMIENTO
plt.imshow(sudoku_gray, "gray")
plt.show()
```

```python
cv.ADAPTIVE_THRESH_GAUSSIAN_C
```

```python
sudoku_adaptive_bin = cv.adaptiveThreshold(sudoku_gray, 255, 
                                           cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, 7, 0)
plt.imshow(sudoku_adaptive_bin, "gray")
plt.title("bs=7 | C=0")
plt.show()
```

```python
bs, c = 11, 2
sudoku_adaptive_bin = cv.adaptiveThreshold(sudoku_gray, 255, 
                                           cv.ADAPTIVE_THRESH_GAUSSIAN_C, cv.THRESH_BINARY, bs, c)
plt.imshow(sudoku_adaptive_bin, "gray")
plt.title(f"GAUSSIAN: bs={bs} | C={c}")
plt.show()
```

```python
bs, c = 11, 2
sudoku_adaptive_bin = cv.adaptiveThreshold(sudoku_gray, 255, 
                                           cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, bs, c)
plt.imshow(sudoku_adaptive_bin, "gray")
plt.title(f"MEAN: bs={bs} | C={c}")
plt.show()
```
